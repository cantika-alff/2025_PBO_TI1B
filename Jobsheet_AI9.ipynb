```python
# import library yang digunakan
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# load dataset
dataset = pd.read_csv('iklan_sosmed.csv', delimiter=';')
dataset.head()

```

```python
X = dataset.iloc[:,2:-1]
y = dataset.iloc[:, -1]

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=99)

X_train = X_train.values
y_train = y_train.values
X_test = X_test.values
y_test = y_test.values

from sklearn.preprocessing import MinMaxScaler
sc = MinMaxScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

from sklearn.naive_bayes import GaussianNB
classifier = GaussianNB()
classifier.fit(X_train, y_train)

```

```python
# prediksi X_test
y_pred = classifier.predict(X_test)

y_pred
y_test
```

```python
# buat confussion matrix & skor akurasi
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
acc = accuracy_score(y_test, y_pred)

print("Akurasi : {}%".format(acc*100))
print("Confussion Matrix : ")
print(cm)

```

```python
# visualisasi sampel data train
from matplotlib.colors import ListedColormap
X_set, y_set = X_train, y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Klasifier (Sampel Training)')
plt.xlabel('Umur')
plt.ylabel('Perkiraan Gaji')
plt.legend()

```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd

# Load dan split data
df = pd.read_csv('diabetes.csv')
X = df[df.columns[:8]]
y = df['Outcome']

# Normalisasi
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)

# Grid Search SVM
param_grid = {
    'C': [0.1, 1, 10, 100],
    'kernel': ['linear', 'rbf', 'poly'],
    'gamma': ['scale', 'auto']
}
grid = GridSearchCV(SVC(), param_grid, cv=5, scoring='accuracy')
grid.fit(X_train, y_train)

print("Best Parameters:", grid.best_params_)
print("Best Accuracy on Training Set:", grid.best_score_)
print("Test Set Accuracy:", grid.score(X_test, y_test))
```

```python
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd

# Load dataset
df = pd.read_csv('diabetes.csv')

# Ambil dua fitur untuk visualisasi
X = df[['Glucose', 'BMI']]
y = df['Outcome']

# Normalisasi
sc = StandardScaler()
X = sc.fit_transform(X)

# Split data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Latih model terbaik (contoh hasil GridSearch)
clf = SVC(C=10, kernel='rbf', gamma='scale')
clf.fit(X_train, y_train)

# Plot decision boundary
from matplotlib.colors import ListedColormap
import numpy as np

X_set, y_set = X_train, y_train
X1, X2 = np.meshgrid(
    np.arange(X_set[:, 0].min()-1, X_set[:, 0].max()+1, 0.01),
    np.arange(X_set[:, 1].min()-1, X_set[:, 1].max()+1, 0.01)
)

plt.figure(figsize=(8,6))
plt.contourf(X1, X2, clf.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             cmap=ListedColormap(('red','green')), alpha=0.6)
plt.scatter(X_set[:,0], X_set[:,1], c=y_set, cmap=ListedColormap(('red','green')))
plt.title('SVM Decision Boundary (Glucose vs BMI)')
plt.xlabel('Glucose')
plt.ylabel('BMI')
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
import pandas as pd
import numpy as np
from sklearn.metrics import mean_squared_error

# Load data
data = pd.read_csv('Salary_Data.csv')
X = np.array(data['YearsExperience'])[:, np.newaxis]
y = np.array(data['Salary'])

# Grid Search SVR
param_grid = {
    'C': [1, 100, 1000],
    'gamma': [0.001, 0.01, 0.1],
    'kernel': ['rbf', 'linear']
}
grid = GridSearchCV(SVR(), param_grid, cv=5, scoring='neg_mean_squared_error')
grid.fit(X, y)

print("Best Parameters:", grid.best_params_)
print("Best Score (Negative MSE):", grid.best_score_)

# Evaluasi model
y_pred = grid.predict(X)
mse = mean_squared_error(y, y_pred)
print("Final MSE:", mse)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVR

# Load data
data = pd.read_csv('Salary_Data.csv')
X = np.array(data['YearsExperience'])[:, np.newaxis]
y = np.array(data['Salary'])

# Model terbaik dari GridSearch
model = SVR(C=1000, gamma=0.1, kernel='rbf')
model.fit(X, y)

# Visualisasi
plt.figure(figsize=(8,6))
plt.scatter(X, y, color='blue', label='Data Asli')
plt.plot(X, model.predict(X), color='red', linewidth=2, label='Regresi SVR')
plt.title('Support Vector Regression')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score, confusion_matrix

# Load data
dataset = pd.read_csv('iklan_sosmed.csv', delimiter=';')
X = dataset.iloc[:, 2:-1].values
y = dataset.iloc[:, -1].values

# Normalisasi
sc = MinMaxScaler()
X = sc.fit_transform(X)

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=99)

# Grid Search Naive Bayes
param_grid = {
    'var_smoothing': np.logspace(-9, -1, 10)
}
grid = GridSearchCV(GaussianNB(), param_grid, cv=5, scoring='accuracy')
grid.fit(X_train, y_train)

# Evaluasi
y_pred = grid.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print("Best Params:", grid.best_params_)
print("Akurasi Baru:", acc*100)
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import MinMaxScaler
from matplotlib.colors import ListedColormap

# Load dataset
dataset = pd.read_csv('iklan_sosmed.csv', delimiter=';')
X = dataset.iloc[:,2:-1].values
y = dataset.iloc[:,-1].values

# Normalisasi
sc = MinMaxScaler()
X = sc.fit_transform(X)

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=99)

# Train Naive Bayes dengan hasil tuning
classifier = GaussianNB(var_smoothing=1e-6)
classifier.fit(X_train, y_train)

# Visualisasi
X_set, y_set = X_train, y_train
X1, X2 = np.meshgrid(np.arange(start=X_set[:,0].min()-1, stop=X_set[:,0].max()+1, step=0.01),
                     np.arange(start=X_set[:,1].min()-1, stop=X_set[:,1].max()+1, step=0.01))
plt.figure(figsize=(8,6))
plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha=0.75, cmap=ListedColormap(('red', 'green')))
plt.scatter(X_set[:,0], X_set[:,1], c=y_set, cmap=ListedColormap(('red', 'green')))
plt.title('Naïve Bayes (Training Data)')
plt.xlabel('Umur (Scaled)')
plt.ylabel('Gaji (Scaled)')
plt.show()
```
