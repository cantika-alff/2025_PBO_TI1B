## Latihan 1

```python
import numpy as np
from scipy import stats
```

```python
# a. Mendapatkan Nilai Mean
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
mean = np.mean(score)
print("Mean:", mean)

# b. Mendapatkan Nilai Median
median = np.median(score)
print("Median:", median)

# c. Mendapatkan Nilai Modus/ Mode
mode = stats.mode(score, keepdims=True)
print("Mode:", mode.mode[0], " (Frekuensi:", mode.count[0], ")")
```

## Latihan 2

```python
# a. Mendapatkan Nilai Range
data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]
range_ = np.max(data) - np.min(data)
print("Range:", range_)

# b. Mendapatkan Nilai Percentile ke-90
data2 = [90, 80, 70, 65, 110, 80, 83, 67, 94]
percentile_90 = np.percentile(data2, 90)
print("Percentile ke-90:", percentile_90)

# c. Mendapatkan Nilai Quartile & IQR
data3 = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]
Q1 = np.percentile(data3, 25)
Q3 = np.percentile(data3, 75)
IQR = Q3 - Q1
print("Q1:", Q1, "Q3:", Q3, "IQR:", IQR)

# d. Mendapatkan Nilai Variance dan Standard Deviation
data4 = [86, 87, 88, 86, 87, 85, 86]
mean4 = np.mean(data4)
var = np.var(data4)
std = np.std(data4)
print("Mean:", mean4, "Variance:", var, "Standard Deviation:", std)
```

### Penugasan 3

```python
# Data ujian
nilai = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99]
nilai_sorted = sorted(nilai)
Q1 = np.percentile(nilai_sorted, 25)
Q3 = np.percentile(nilai_sorted, 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = [x for x in nilai_sorted if x < lower_bound or x > upper_bound]


print("a. Data urut:", nilai_sorted)
print("b. Q1:", Q1, "Q3:", Q3, "IQR:", IQR)
print("c. Lower Bound:", lower_bound, "Upper Bound:", upper_bound)
print("d. Outliers:", outliers)
```

### Penugasan 4

```python
# Standar deviasi dari data kelompok
# Data berat badan:
# Interval dan frekuensi
intervals = [(41, 45), (46, 50), (50, 55), (56, 60), (61, 65), (66, 70)]
frequencies = [2, 3, 6, 4, 5, 2]

# Menghitung nilai tengah dan total data
midpoints = [(low + high) / 2 for low, high in intervals]
total_freq = sum(frequencies)
mean_grouped = sum([f * m for f, m in zip(frequencies, midpoints)]) / total_freq

# Varians
variance_grouped = sum([f * ((m - mean_grouped) ** 2) for f, m in zip(frequencies, midpoints)]) / total_freq
std_grouped = np.sqrt(variance_grouped)

print("\nStandar Deviasi Data Kelompok")
print("Standar Deviasi:", round(std_grouped, 2))
```
