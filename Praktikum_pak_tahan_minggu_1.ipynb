```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter sinyal
f = 1   # frekuensi 1 Hz
t = np.linspace(0, 2*np.pi, 1000)   # waktu kontinyu
n = np.arange(0, 10, 1)             # indeks diskrit

# Definisi sinyal
x_t = np.sin(t)    # sinyal kontinyu
x_n = np.sin(n)    # sinyal diskrit

# Plot
plt.figure(figsize=(10,4))

# --- Sinyal Kontinyu ---
plt.subplot(1,2,1)
plt.plot(t, x_t, 'b')
plt.title("Sinyal Kontinyu x(t)")
plt.xlabel("t")
plt.ylabel("x(t)")
plt.grid(True)

# --- Sinyal Diskrit ---
plt.subplot(1,2,2)
plt.stem(n, x_n, linefmt='b-', markerfmt='bo', basefmt='k')
plt.title("Sinyal Diskrit x[n]")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.grid(True)

plt.tight_layout()
plt.show()
```
